```python
import pandas as pd
import numpy as np
print("Imported")
```

```python
dict_df = {
    "id"          : [1        ,2         ,3              ,4            ,5        ,106         ],
    "name"        : ["ariel"  ,"nadav"   ,"maor"         ,"noam"       ,"lior"   ,"hamarmir"],
    "age"         : [25       , 24       ,26             ,26           ,None       ,45        ],
    "healthy"     : [True     ,True      ,False          ,False        ,True     ,True      ],
    "saving_count": [1200     ,1500      ,2140           ,1523         ,3000     ,2500      ],
    "country"     : ["israel" ,"israel"  ,'north_korea'  ,"thailand"   ,'peru'   ,'greece'  ]
}

df = pd.DataFrame(dict_df,
                  columns=list(dict_df.keys()))

display(df)
```

```python
df_dates = pd.DataFrame(
    {'sales': [100, 150, 200]},
    index=pd.to_datetime([
        '2026-01-01',
        '2026-01-02',
        '2026-01-03'
    ])
)
display(df_dates)
new_df_dates = df_dates.shift(freq='MS',axis=0,periods=10)
display(new_df_dates)
print(f"\r{'='*50}")
display(df)
new_df = df.shift(axis=0,periods=1)
#Copy with values lagged by 1.
### (periods=1,freq=None,axis=0,fill_value=None,suffix=None)
## periods = 1 -> down, = -1 -> up.
## fill_value=0 -> might be problem for data type.
## freq='D'/'h'/'min'/'MS'

#print("df.shift(-1)")
display(new_df)
```

```python
new_df = df['age'].cumsum()
#Cumulative sum.
### (axis=0,skipna=True)
## skipna=True -> Default, will skip the NaN Values, it will fill the NaN index with NaN value and continue normal.
# skipna=False -> will fill each of the next fields with NaN values.

print("df['age']\n",df['age'],'df["age"].cumsum()\n',new_df)
```

```python
new_df = df['age'].cummax()
#Cumulative max.
### (axis=0,skipna=True)

print("df['age']\n",df['age'],"df['age'].cummax()\n",new_df)
```

```python
new_df = df['age'].cummin()
#Cumulative min.

print("df['age']\n",df['age'],"df['age'].cummin()\n",new_df)
```

```python
new_df = df['age'].cumprod()
#Cumulative product.
print("df['age']\n",df['age'],"df['age'].cumprod()\n",new_df)
```

```python
new_df = df[['age','country']].rank(method='dense',numeric_only=True,pct=True)
#Ranks. Ties go to first value.
### (axis=0,method='average',numeric_only=False,na_option='keep',ascending=True,pct=False)
## method=['first','average','min','max','dense']
# method='first' -> first will get the higher rank.
# method = 'average' -> both same values will get their average rank.
# method = 'min' -> both same values will get the min value of both.
# method = 'max' -> both same values will get the max value of both.
# method = 'dense' -> both same values will get the same rank and next rank keep growing normally.
## na_option= ['keep','top','bottom']
## pct=True -> return the rank represented as percents.

print("df[['age','country']]\n",df[['age','country']],f"\n{'='*100}\ndf[['age','country']].rank(method='average',numeric_only=True,ascending=True)\n",new_df)
```
