```python
import pandas as pd

print("Imported")
```

```python
dict_df = {
    "id"          : [1        ,2         ,3         ,4],
    "name"        : ["ariel"  ,"nadav"   ,"meir"    ,"noam"],
    "age"         : [25       , 24       ,26        ,23],
    "healthy"     : [True     ,True      ,False     ,False],
    "saving_count": [1200     ,1500      ,2140      ,1523],
    "country"     : ["israel" ,"israel"  ,"yarden"  ,"thailand"]
}
df = pd.DataFrame(dict_df,
                  columns=list(dict_df.keys()))

display(df)
```

```python
new_df = df.sort_values(by=['name','saving_count'], ascending=[False, True],inplace=False,
                        na_position='last',kind="quicksort",ignore_index=True)

### (by="mpg",ascending=True,inplace=False,na_position="last",kind="quicksort",ignore_index=False,key=None)

# Order rows by values of a column (low to high).
## by -> if list of columns it is gonna sort the df by the first element if there is equal
# elements in 2 different indexes it will sort by the next element of by=.
## ascending = False -> from lower to higher. ascending = False -> from higher to lower.
## inplace = False -> Default, send a new sorted *copy* , ...=True -> change the original object(return None).
## na_position='last' -> determine where the NaN Values will be sorted to(last/first).
## kind="quicksort" -> determine the sort method,(mergesort->more stable,heapsort->less common)
## ignore_index=True -> reset the indexes .
## key -> lambda function. lambda input -> Series('name') before the sort.

display(new_df)

```

```python
new_df = df.rename(columns = {'name':'eman'},
                   index={0: 'zero', 1: 'one'},
                   inplace=False,
                   errors='ignore',)
# Rename the columns of a DataFrame only changing the metadata.
## errors='ignore' -> Default ignore the error if not key found.(raise is the negate for ignore)

display(new_df)
```

```python
df2 = pd.DataFrame(dict_df,index=[5,1,4,7])
display(df2)
new_df1 = df2.sort_index()
new_df2 = df2.sort_index(ignore_index=True)

# Sort the index of a DataFrame
### (axis=0,ascending=True,inplace=False,kind='quicksort'
### ,na_position='last',sort_remaining=True,ignore_index=False,key=None)
## axis = 0 -> Default, sort by index of rows. axis = 1 -> sort by column name.
## ignore_index = True -> reset the index
display(new_df1)
display(new_df2)
```

```python
df = pd.DataFrame(dict_df)
display(df)
melt_df = pd.melt(df,id_vars=['id','name'],value_vars=['saving_count','healthy','country','age'],var_name='column',
                 value_name='column_value',ignore_index=False)
#Gather columns into rows.
# if you mention both id_vars and value_var it will drop the columns that have not mentioned.
## problem with time complexity, when you not have between 100k-500k rows and only tens of columns
display(melt_df)

```

```python
dict_df = {
    "id"          : [1        ,2         ,3              ,4            ,5        ,106       ],
    "name"        : ["ariel"  ,"nadav"   ,"maor"         ,"noam"       ,"lior"   ,"hamarmir"],
    "age"         : [25       , 24       ,26             ,26           ,None     ,45        ],
    "healthy"     : [True     ,True      ,False          ,False        ,True     ,True      ],
    "saving_count": [1200     ,1500      ,2140           ,1523         ,3000     ,2500      ],
    "addicted"    : ["high"   ,"medium"  ,"high"         ,"medium"     ,"less"   ,"not"],
    "country"     : ["israel" ,"israel"  ,'north_korea'  ,"thailand"   ,'peru'   ,'greece'  ]
}

df = pd.DataFrame(dict_df,
                  columns=list(dict_df.keys()))

df1 = df.iloc[:,[0,1,2,3]]
df2 = df.iloc[:4,[0,4,5,6]]
display(df1,df2)
merged_df = pd.merge(
    left=df1,
    right=df2,
    on=['id'],
    how="outer",
    indicator=True,
    validate="one_to_one"
)
### (left,right,how="inner",on=None,left_on=None,right_on=None,left_index=False,right_index=False,sort=False,suffixes=("_x", "_y"),
###    copy=None,indicator=False,validate=None)
## how = 'inner' -> keys that exists in both tables
## how = 'left'  -> keys that exists in left table.
## how = 'right' -> keys that exists in right table.
## how = 'outer' -> all keys from both tables(missing values in one table will be filled with NaN)
## how = 'cross' -> each row from left table will be served with all rows from the right table.(remove 'on' hp)
## on  -> you can choose the columns that exists in both tables, the merged will be considered by their priority order.
## left_on=None,right_on=None -> here you can mention specific column to compare with the other table.(in case of unrelated names)
## left_index=False,right_index=False -> use the right/left table index as his 'on' column.
## sort=False -> sort by keys or not sorted(Default)
## suffixes=("_x", "_y") -> if there is same column names but not in 'on', it will rename 'score'_x for left 'score'_y for right.
## suffixes=(None, "_y") -> will rename only the right table.
## indicator=True -> will return one more column: '_merge', this column use is to explain in what table the row exists.(you can name it)
## validate="one_to_one"  -> will allow to the same key to return only 1 row else ERROR.
## validate="one_to_many" -> will allow to the same key to return more than 1 row only for right table.
## validate="many_to_one" -> will allow only for left table to return more than 1 row only for left table.
## validate="many_to_many"-> will allow  for both tables to return more than 1 row.

display(merged_df)
```
