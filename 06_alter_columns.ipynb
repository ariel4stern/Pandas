```python
import pandas as pd
import numpy as np
import time


print("Imported")
```

```python
dict_df = {
    "id"          : [1        ,2         ,3              ,4            ,5        ,106         ],
    "name"        : ["ariel"  ,"nadav"   ,"maor"         ,"noam"       ,"lior"   ,"hamarmir"],
    "age"         : [25       , 24       ,26             ,23           ,21       ,45        ],
    "healthy"     : [True     ,True      ,False          ,False        ,True     ,True      ],
    "saving_count": [1200     ,1500      ,2140           ,1523         ,3000     ,2500      ],
    "country"     : ["israel" ,"israel"  ,'north_korea'  ,"thailand"   ,'peru'   ,'greece'  ]
}

df = pd.DataFrame(dict_df,
                  columns=list(dict_df.keys()))

display(df)
```

```python
df.assign(**{
          "name_changing" : lambda series: series['name'].where(series['name'] != 'noam' , None),
          "age_in_two_years":df['age']+2,
          "age_in_three_years":lambda series: series['age_in_two_years']+1,
          "column_adding_time":time.strftime("%H:%M:%S---%d/%m/%y"),
          "positive":True
})

# Compute and append one or more new columns.
# with 'lambda' you can mention the column you are building during the assign function.
# if the name of column already exists the new column will replace the column.
# the assign function know to sort when you send series for new column by the series indexes
# it is matching index for index.
```

```python
groups , bins = pd.qcut(df['id'], q=3, labels=[33,66,99],retbins=True,precision=2)
print(f"groups: {groups}.\n\rbins: {bins}")
df['new_qcut'] = pd.qcut(df['id'], q=3, labels=[33,66,99],precision=2)
# qcut will disperse the values to q groups never-mind the range.

groups , bins = pd.cut(df['id'], bins=3, labels=[33,66,99],retbins=True,precision=2)
print(f"groups: {groups}.\n\rbins: {bins}")
df['new_cut'] = pd.cut(df['id'], bins=3, labels=[33,66,99],precision=2)
# cut will made value range and cluster the groups by 1/q of the range each.

### pd.qcut(x,q,labels=None,retbins=False,precision=3,duplicates='drop')
## x -> values to cluster
## q -> the desired precents of groups.
## labels=[33,66,99] -> for q=3;  q1 = 33 , q2 = 66 , q3 = 99.
## duplicates='drop' -> when you have duplicates separate boundary it will drop the duplicated boundary.
## duplicates='drop' -> (not changing the df)
## duplicates='drop' -> raise ERROR if duplicate boundary.

display(df)
```
