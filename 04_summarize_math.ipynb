```python
dict_df = {
    "id"          : [1        ,2         ,3              ,4            ,5        ],
    "name"        : ["ariel"  ,"nadav"   ,"maor"         ,"noam"       ,"lior"   ],
    "age"         : [25       , 24       ,26             ,23           ,21       ],
    "healthy"     : [True     ,True      ,False          ,False        ,True     ],
    "saving_count": [1200     ,1500      ,2140           ,1523         ,3000     ],
    "country"     : ["israel" ,"israel"  ,'north_korea'  ,"thailand"   ,'peru'   ]
}
df = pd.DataFrame(dict_df,
                  columns=list(dict_df.keys()))

display(df)
```

```python
numeric_cols = df.select_dtypes(include='number').drop(columns=['id'])
display(numeric_cols)
min_df = numeric_cols.min()
#Minimum value in each object.
print(min_df)
```

```python
display(numeric_cols)

mean_df = numeric_cols.mean(axis=0)
#Mean value of each object
print(mean_df)
```

```python
display(numeric_cols)

var_df = numeric_cols.var(axis=0)
#Variance of each object.
print(var_df)


import numpy as np

mean_age = numeric_cols['age'].mean()
after_sub_power = np.power([value-mean_age for value in numeric_cols['age']],2)
print(f'\n\r{after_sub_power}')
sum_after_sub = np.sum(after_sub_power)
print(sum_after_sub)
result_var_age = np.divide(sum_after_sub,(len(numeric_cols['age'])-1))
print("age var:",result_var_age)
```

```python
std_df = numeric_cols['age'].std()
# Standard deviation of each object.
print(std_df)

age_seris = numeric_cols['age']

m = np.mean(age_seris)
st = np.subtract(age_seris, m)
power_st = np.power(st ,2)
m_power_st = np.sum(power_st)/(len(age_seris)-1) # (len(age_seris)-1) not in np.mean()
print("age std: ",np.sqrt(m_power_st))
```

```python
q = numeric_cols.quantile(q=[0.25,0.75])
print('numeric_cols.quantile(q=[0.25,0.75])\n',q)
#Quantiles of each object. starting a values range for each numeric column,then you enter the precent you want to see.
###(q=0.5,axis=0,interpolation='linear',method='single')
## q -> the precents you want to calculate.(0-1)
## interpolation            -> the calculating of the missing value.
## interpolation='linear'   -> Proportional calculation between values.
## interpolation='midpoint' -> mean of both nearest values.
## interpolation='nearest'  -> choosing the nearest value to the calculated precents.
## interpolation='higher'   -> higher value chosen.
## interpolation='lower'    -> lower value chosen.
## method='single'          -> Default. calculating each column separately.
## method='table'           -> calculating the all table with relations between columns(lower,higher,nearest).

Q1 = q.loc[0.25]                ## represent of the value in the 25% of the values range
Q3 = q.loc[0.75]                ## represent of the value in the 75% of the values range
IQR = Q3 - Q1

lower_boundary = Q1-1.5*IQR     ## represent the min edge of the normal value,all lower values in to continue will be outliers.
print(f"\nLower Boundary:\n\r{lower_boundary}")

upper_boundary = Q3+1.5*IQR     ## represent the max edge of the normal value,all higher values will be outliers.
print(f"\nUpper Boundary:\n\r{upper_boundary}")
```

```python
display(df)
df['age'] = df['age'].apply(lambda x: x + 10)
display(df)
#Apply function to each object.

df['saving_count'] = df['saving_count'].apply(lambda x: max(x,max(df['saving_count'])))
display(df)

df['country'] = df['country'].apply(
    lambda country:  None if country == 'israel' else country)
display(df)

def add_number(series:pd.Series,number:int):
    return series+number

df['id'] = df['id'].apply(add_number,number=20)
display(df)
```
