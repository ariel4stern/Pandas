```python
import pandas as pd
print("Imported")
```

```python
dict_df = {
    "id"          : [2        ,2         ,2         ,4],
    "name"        : ["ariel"  ,"nadav"   ,"ariel"    ,"noam"],
    "age"         : [25       , 24       ,26        ,23],
    "healthy"     : [True     ,True      ,False     ,False],
    "saving_count": [1200     ,1500      ,2140      ,1523],
    "country"     : ["israel" ,"israel"  ,"yarden"  ,"thailand"]
}
df = pd.DataFrame(dict_df,
                  columns=list(dict_df.keys()))

display(df)
```

```python
print("DataFrame")
display(df)

after_drop = df.drop_duplicates(subset=["name","id"],keep=False,inplace=False,ignore_index=True)
#Remove duplicate rows (only considers columns).
### (subset=None,*,keep="first",inplace=False,ignore_index=False)
## subset=["name"] -> focusing the drop to be applied on rows with duplicate specific columns.(all(subset))
## keep = ["first","last",False] ->
# "first" -> first in order stay.
# "last" -> last in order stay.
# False -> drop all duplicate rows. nothing stay
## inplace -> return None, and changing the original dataframe.(Default=False)
## ignore_index=True -> reindexing the dateframe again from 0.
print("df.drop_duplicates()")
display(after_drop)
```

```python
print("DateFrame")
display(df)
selected = df.sample(n= len(df),replace=True,random_state=42)
#Randomly select fraction of rows.
###(n=None,frac=None,replace=False,weights=None,random_state=None,axis=None,ignore_index=False)
## frac=0.5 -> 50% of the dataframe rows will be selected to return.(WARNING: n and frac hyperparameters can't be used together)
## n=2 -> 2 random samples to return.(WARNING: n and frac hyperparameters can't be used together)
## replace=False -> Default, can't choose the same row more than 1 time. ...=True the same row might choose more than 1 time.


print("df.sample(frac=0.5)")
display(selected)
```

```python
print("Dataframe")
display(df)
selected = df.nlargest(n = 2, columns=['id'],keep="all")
#Select and order top n entries.
### (n,columns,keep='first')
## keep="all" -> in case there is tie and the number of selected rows is higher than
# the *n* hyperparameter so the number of selected rows will be all values with max values.

print("df.nlargest(n, 'value')")
display(selected)
```

```python
print("Dataframe")
display(df)
selected = df.nsmallest(n = 2, columns=['id'],keep='last')
#Select and order bottom n entries.
### (n,columns,keep='first')
## keep="all" -> in case there is tie and the number of selected rows is higher than
# the *n* hyperparameter so the number of selected rows will be all values with max values.

print("df.nsmallest(n = 2, columns=['id'],keep='last')")
display(selected)
```
