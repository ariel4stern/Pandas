```python
import pandas as pd
import numpy as np
print("Imported")
```

```python
dict_df = {
    "id": [1, 2, 3, 4, 5],
    "name": ["ariel", "nadav", "maor", "noam", "lior"],
    "age": [25., None, 26., 23., 21.],
    "healthy": [True, True, False, False, None],
    "saving_count": [1200., 1500., np.nan, None, 3000.],
    "country": ["israel", None, 'peru', None, 'peru']
}
df = pd.DataFrame(dict_df,
                  columns=list(dict_df.keys()))

display(df)
```

```python
clean_df = df.dropna(axis=1,subset=[0,1,2],how='any')
# Drop rows with any column having None(python missing)/NaN(number missing)/NaT(time missing) data.
# Default-> removing the row with the missing value.
### (axis=0,how='any',thresh=None,subset=None,inplace=False,ignore_index=False)
## axis = 0 -> rows .axis = 1-> columns.
## how='any' -> if there is one missing value in any column -- drop row (default).
## how='all' -> if there is missing value in all columns -- drop row.
## thresh -> if there is thresh count of existing values in the axis, it will be safe.
## subset -> set the specific columns/rows you desire.

display(clean_df)
```

```python
filled_df = df.fillna(value=({
    'age': np.median(df.dropna(axis=0,how='any')['age']),
    'healthy': False,
    'saving_count': np.mean(df['saving_count']),
    'country': df['country'].mode()[0]
}))
# Replace all NA/null data with value.
###( value=None,axis=None,inplace=False,limit=None)
## value -> the values to fill the missing value,(int,str,boll,dict[column])
## limit -> will keep dropping until the limit value over.(limit=2, only 2 values will be filled)

display(filled_df)

display(df['age'],df['age'].ffill())
## fill by the previews value.

print('='*50)
display(df['age'],df['age'].bfill())
## fill by the next value.

```
